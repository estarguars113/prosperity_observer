# Prosperity Indicators – Data Exploration

This notebook demonstrates how to interactively explore the prosperity lakehouse using PySpark and Delta Lake.

```python
from prosperity_processing.pyspark_session import build_spark_session
from prosperity_processing.reader import read_delta_table

spark = build_spark_session()
df = read_delta_table(spark, "../../extract/lakehouse/transactions")
df.printSchema()
```

```python
# Show sample data
df.show(5, truncate=False)
```

```python
# Basic statistics per indicator
from pyspark.sql import functions as F

stats = df.groupBy("indicator_id").agg(
    F.count("*").alias("count"),
    F.avg("value").alias("avg_value"),
    F.min("year").alias("min_year"),
    F.max("year").alias("max_year"),
)
stats.toPandas()
```

```python
# Data validation check
from prosperity_processing.validation import validate_indicator_df

result = validate_indicator_df(df)
print("Validation success:", result["success"])
print("Statistics:", result["statistics"])
```
